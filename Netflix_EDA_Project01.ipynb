```python
import pandas as pd

df = pd.read_csv('netflix_titles.csv')
print(df.shape)
df.head()
```

```python
print(df.isnull().sum())
```

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')

df = df.dropna(subset=['date_added', 'rating', 'duration'])

print(df.isnull().sum())
print("\nDataset shape after cleaning:", df.shape)
```

```python

print("""
Dataset Summary:
1. The Netflix dataset contains 8807 titles (Movies and TV Shows)
2. It has 12 columns including title, type, director, cast, country, rating and duration
3. Director column had the most missing values (2634) - filled with 'Unknown'
4. After cleaning, rows with missing date/rating/duration were removed
5. The cleaned dataset is ready for Exploratory Data Analysis
""")
```

```python
type_counts = df['type'].value_counts()
print(type_counts)
```

```python
top_countries = df['country'].value_counts().head(10)
print(top_countries)
```

```python
rating_counts = df['rating'].value_counts()
print(rating_counts)
```

```python
df['year_added'] = df['date_added'].str.strip().str[-4:]
yearly_counts = df['year_added'].value_counts().sort_index()
print(yearly_counts)
```

```python
genres = df['listed_in'].str.split(', ').explode()
top_genres = genres.value_counts().head(10)
print(top_genres)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart 1: Movies vs TV Shows (Pie Chart)
plt.figure(figsize=(6,6))
df['type'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['#E50914','#221F1F'])
plt.title('Movies vs TV Shows on Netflix')
plt.ylabel('')
plt.savefig('chart1_pie.png')
plt.show()

# Chart 2: Top 10 Countries (Bar Chart)
plt.figure(figsize=(10,5))
df['country'].value_counts().head(10).plot(kind='bar', color='#E50914')
plt.title('Top 10 Countries with Most Netflix Content')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('chart2_bar.png')
plt.show()

# Chart 3: Ratings Distribution (Bar Chart)
plt.figure(figsize=(10,5))
df['rating'].value_counts().plot(kind='bar', color='#221F1F')
plt.title('Netflix Content by Rating')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('chart3_ratings.png')
plt.show()

# Chart 4: Content Added Per Year (Line Chart)
plt.figure(figsize=(10,5))
df['year_added'].value_counts().sort_index().plot(kind='line', marker='o', color='#E50914')
plt.title('Netflix Content Added Per Year')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.tight_layout()
plt.savefig('chart4_line.png')
plt.show()

# Chart 5: Top 10 Genres (Horizontal Bar)
plt.figure(figsize=(10,5))
genres = df['listed_in'].str.split(', ').explode()
genres.value_counts().head(10).plot(kind='barh', color='#E50914')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Count')
plt.tight_layout()
plt.savefig('chart5_genres.png')
plt.show()

# Chart 6: Movies vs TV Shows per Year (Stacked Bar)
plt.figure(figsize=(12,5))
year_type = df.groupby(['year_added','type']).size().unstack()
year_type.plot(kind='bar', stacked=True, color=['#E50914','#221F1F'])
plt.title('Movies vs TV Shows Added Per Year')
plt.xlabel('Year')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('chart6_stacked.png')
plt.show()
```
